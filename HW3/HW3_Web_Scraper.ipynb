```python
import json
from bs4 import BeautifulSoup
import requests
```

```python
links = []
url = "https://bama.ir/car/samand?year=1385,1401"
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
    'Accept-Language': 'en-US,en;q=0.9',
}
# This function crawls the page and saves links that have news with comments.

def crawler():
        url = f"https://bama.ir/car/samand?year=1385,1401"
        response = requests.get(url,headers=headers)
        soup = BeautifulSoup(response.text, 'html.parser')
        cars = soup.findAll('div', {'class': 'bama-ad-holder'})
    
        for car in cars:
            re = "https://bama.ir" + car.find('a')['href']
            print(re)
            links.append(re)

crawler()

# At the end, we save the news link in the form of a json file.

with open("links.json", "a") as outfile:
    json.dump(links, outfile)
outfile.close()
```

```python
cars_data=[]
for i in range(15):
    response=requests.get(f"https://bama.ir/cad/api/search?yearFrom=1385&yearTo=1401&vehicle=samand&pageIndex={i}", headers=headers)
    cars=response.json()["data"]["ads"]
    for cars1 in cars:
        data={
            "title":cars1["detail"]["title"],
            "url":cars1["detail"]["url"],
            "mileage":cars1["detail"]["mileage"],
            "color":cars1["detail"]["color"],
            "description":cars1["detail"]["description"],
            "transmission":cars1["detail"]["transmission"],
            "year":cars1["detail"]["year"],
            "price":cars1["price"]["price"],
        }
        cars_data.append(data)        
       
```
